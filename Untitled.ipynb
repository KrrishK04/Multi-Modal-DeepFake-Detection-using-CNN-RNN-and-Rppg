```python
from __future__ import annotations

import os
import csv
import json
import math
import shutil
import subprocess
from pathlib import Path
from dataclasses import dataclass
from typing import List, Dict, Tuple, Optional
from concurrent.futures import ProcessPoolExecutor, as_completed

import pandas as pd
from tqdm.auto import tqdm
```

```python
# -------- USER CONFIG --------
FFPP_ROOT = Path("./FaceForensics++_C23").resolve()
METADATA_CSV = FFPP_ROOT / "csv" / "FF++_Metadata.csv"

OUTPUT_ROOT = Path("./preprocessed_ffpp").resolve()
FRAMES_ROOT = OUTPUT_ROOT / "frames"
MANIFEST_CSV = OUTPUT_ROOT / "clips_manifest.csv"
PREPROCESS_META_JSON = OUTPUT_ROOT / "preprocess_meta.json"

# Sampling / image settings
TARGET_FPS = 8              # extract 8 frames per second
JPEG_QUALITY = 3            # ffmpeg qscale:v ; lower is better quality
RESIZE_SHORTER_SIDE = 256   # resize during preprocessing; train can still center/resize to 224
KEEP_EXISTING = True        # skip videos already processed
MAX_VIDEOS = 0              # 0 = all

# Parallelism for MacBook
MAX_WORKERS = 4             # good starting point on MacBook; raise cautiously
CHUNKSIZE = 8

# Optional filtering
CATEGORIES = ""             # e.g. "original,Deepfakes"
SKIP_MISSING = True
```

```python
def ensure_dirs():
    OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
    FRAMES_ROOT.mkdir(parents=True, exist_ok=True)

def run_cmd(cmd: list[str]) -> subprocess.CompletedProcess:
    return subprocess.run(
        cmd,
        stdout=subprocess.PIPE,
        stderr=subprocess.PIPE,
        text=True,
        check=False,
    )

def ffmpeg_exists() -> bool:
    result = run_cmd(["ffmpeg", "-version"])
    return result.returncode == 0

def ffprobe_exists() -> bool:
    result = run_cmd(["ffprobe", "-version"])
    return result.returncode == 0

def safe_video_id(rel_path: str) -> str:
    return rel_path.replace("\\", "/").replace("/", "__").rsplit(".", 1)[0]

def count_jpgs(folder: Path) -> int:
    return len(list(folder.glob("*.jpg")))

def load_metadata_rows(csv_path: Path) -> List[Dict]:
    # expects columns compatible with your FF++ metadata helper:
    # rel_path, label, category
    rows = []
    with open(csv_path, "r", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for row in reader:
            rows.append({
                "rel_path": row["File Path"] if "File Path" in row else row.get("rel_path", ""),
                "label": row["Label"] if "Label" in row else row.get("label", ""),
                "category": row.get("category", ""),
            })
    return rows
```

```python
def collect_videos(
    ffpp_root: Path,
    metadata_csv: Path,
    categories: str = "",
    skip_missing: bool = True,
    max_videos: int = 0,
) -> List[Dict]:
    rows = load_metadata_rows(metadata_csv)

    cats_filter = None
    if categories.strip():
        cats_filter = {x.strip() for x in categories.split(",") if x.strip()}

    items = []
    skipped_missing = 0

    for row in rows:
        rel_path = row["rel_path"]
        label = row["label"].strip().upper()
        category = row["category"]

        if cats_filter is not None and category not in cats_filter:
            continue

        full_path = (ffpp_root / rel_path).resolve()

        if not full_path.is_file():
            if skip_missing:
                skipped_missing += 1
                continue
            raise FileNotFoundError(full_path)

        if label not in {"FAKE", "REAL"}:
            continue

        items.append({
            "rel_path": rel_path,
            "full_path": str(full_path),
            "label": 0 if label == "FAKE" else 1,
            "label_name": label,
            "category": category,
            "video_id": safe_video_id(rel_path),
        })

    if max_videos > 0:
        items = items[:max_videos]

    print(f"Collected {len(items)} videos")
    if skipped_missing:
        print(f"Skipped missing: {skipped_missing}")

    return items
```

```python
@dataclass
class ProcessResult:
    video_id: str
    rel_path: str
    label: int
    label_name: str
    category: str
    frames_dir: str
    num_frames: int
    ok: bool
    error: str = ""

def process_one_video(item: Dict) -> ProcessResult:
    video_id = item["video_id"]
    rel_path = item["rel_path"]
    label = item["label"]
    label_name = item["label_name"]
    category = item["category"]
    src = Path(item["full_path"])
    out_dir = FRAMES_ROOT / video_id

    try:
        if KEEP_EXISTING and out_dir.exists():
            existing = count_jpgs(out_dir)
            if existing > 0:
                return ProcessResult(
                    video_id=video_id,
                    rel_path=rel_path,
                    label=label,
                    label_name=label_name,
                    category=category,
                    frames_dir=str(out_dir),
                    num_frames=existing,
                    ok=True,
                )

        if out_dir.exists():
            shutil.rmtree(out_dir)
        out_dir.mkdir(parents=True, exist_ok=True)

        # scale so shorter side = RESIZE_SHORTER_SIDE, preserving aspect ratio
        vf = (
            f"fps={TARGET_FPS},"
            f"scale='if(lt(iw,ih),{RESIZE_SHORTER_SIDE},-2)':'if(lt(iw,ih),-2,{RESIZE_SHORTER_SIDE})'"
        )

        cmd = [
            "ffmpeg",
            "-hide_banner",
            "-loglevel", "error",
            "-i", str(src),
            "-vf", vf,
            "-qscale:v", str(JPEG_QUALITY),
            str(out_dir / "%06d.jpg"),
        ]

        result = run_cmd(cmd)
        if result.returncode != 0:
            return ProcessResult(
                video_id=video_id,
                rel_path=rel_path,
                label=label,
                label_name=label_name,
                category=category,
                frames_dir=str(out_dir),
                num_frames=0,
                ok=False,
                error=result.stderr.strip(),
            )

        num_frames = count_jpgs(out_dir)
        if num_frames == 0:
            return ProcessResult(
                video_id=video_id,
                rel_path=rel_path,
                label=label,
                label_name=label_name,
                category=category,
                frames_dir=str(out_dir),
                num_frames=0,
                ok=False,
                error="No frames extracted",
            )

        return ProcessResult(
            video_id=video_id,
            rel_path=rel_path,
            label=label,
            label_name=label_name,
            category=category,
            frames_dir=str(out_dir),
            num_frames=num_frames,
            ok=True,
        )

    except Exception as e:
        return ProcessResult(
            video_id=video_id,
            rel_path=rel_path,
            label=label,
            label_name=label_name,
            category=category,
            frames_dir=str(out_dir),
            num_frames=0,
            ok=False,
            error=str(e),
        )
```

```python
ensure_dirs()

assert ffmpeg_exists(), "ffmpeg not found on PATH"
assert ffprobe_exists(), "ffprobe not found on PATH"
assert FFPP_ROOT.exists(), f"Missing FFPP root: {FFPP_ROOT}"
assert METADATA_CSV.exists(), f"Missing metadata CSV: {METADATA_CSV}"

items = collect_videos(
    ffpp_root=FFPP_ROOT,
    metadata_csv=METADATA_CSV,
    categories=CATEGORIES,
    skip_missing=SKIP_MISSING,
    max_videos=MAX_VIDEOS,
)

results: List[ProcessResult] = []
failures: List[ProcessResult] = []

with ProcessPoolExecutor(max_workers=MAX_WORKERS) as ex:
    futures = [ex.submit(process_one_video, item) for item in items]

    for fut in tqdm(as_completed(futures), total=len(futures), desc="Preprocessing"):
        res = fut.result()
        if res.ok:
            results.append(res)
        else:
            failures.append(res)

print(f"Success: {len(results)}")
print(f"Failed : {len(failures)}")
```

```python
rows = []
for r in results:
    rows.append({
        "video_id": r.video_id,
        "rel_path": r.rel_path,
        "label": r.label,
        "label_name": r.label_name,
        "category": r.category,
        "frames_dir": r.frames_dir,
        "num_frames": r.num_frames,
        "split": "",   # fill later once you create fixed train/val/test splits
    })

df = pd.DataFrame(rows).sort_values("video_id").reset_index(drop=True)
df.to_csv(MANIFEST_CSV, index=False)

meta = {
    "ffpp_root": str(FFPP_ROOT),
    "metadata_csv": str(METADATA_CSV),
    "output_root": str(OUTPUT_ROOT),
    "target_fps": TARGET_FPS,
    "jpeg_quality": JPEG_QUALITY,
    "resize_shorter_side": RESIZE_SHORTER_SIDE,
    "max_workers": MAX_WORKERS,
    "num_processed": len(results),
    "num_failed": len(failures),
}
with open(PREPROCESS_META_JSON, "w", encoding="utf-8") as f:
    json.dump(meta, f, indent=2)

print(f"Saved manifest: {MANIFEST_CSV}")
print(f"Saved meta    : {PREPROCESS_META_JSON}")
```

```python
from sklearn.model_selection import train_test_split

VAL_FRACTION = 0.15
TEST_FRACTION = 0.15
SEED = 42

manifest = pd.read_csv(MANIFEST_CSV)

train_val_df, test_df = train_test_split(
    manifest,
    test_size=TEST_FRACTION,
    random_state=SEED,
    stratify=manifest["label"],
)

val_ratio_of_remainder = VAL_FRACTION / (1.0 - TEST_FRACTION)

train_df, val_df = train_test_split(
    train_val_df,
    test_size=val_ratio_of_remainder,
    random_state=SEED,
    stratify=train_val_df["label"],
)

train_df = train_df.copy()
val_df = val_df.copy()
test_df = test_df.copy()

train_df["split"] = "train"
val_df["split"] = "val"
test_df["split"] = "test"

full_df = pd.concat([train_df, val_df, test_df], ignore_index=True)
full_df.to_csv(MANIFEST_CSV, index=False)

TRAIN_SPLIT_CSV = OUTPUT_ROOT / "train_split.csv"
VAL_SPLIT_CSV = OUTPUT_ROOT / "val_split.csv"
TEST_SPLIT_CSV = OUTPUT_ROOT / "test_split.csv"

train_df.to_csv(TRAIN_SPLIT_CSV, index=False)
val_df.to_csv(VAL_SPLIT_CSV, index=False)
test_df.to_csv(TEST_SPLIT_CSV, index=False)

print(TRAIN_SPLIT_CSV)
print(VAL_SPLIT_CSV)
print(TEST_SPLIT_CSV)
```

```python
import random
import numpy as np
from PIL import Image
import torch
from torch.utils.data import Dataset
import torchvision.transforms as transforms

def uniform_sample_indices(n: int, clip_len: int):
    if n <= 0:
        raise ValueError("No frames")
    if n >= clip_len:
        return np.linspace(0, n - 1, clip_len, dtype=int).tolist()
    idx = np.linspace(0, n - 1, clip_len)
    return np.round(idx).astype(int).tolist()

class CachedFramesDataset(Dataset):
    def __init__(self, csv_path: str, clip_len: int, transform):
        self.df = pd.read_csv(csv_path).reset_index(drop=True)
        self.clip_len = clip_len
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        frames_dir = Path(row["frames_dir"])
        label = int(row["label"])

        frame_paths = sorted(frames_dir.glob("*.jpg"))
        if len(frame_paths) == 0:
            raise ValueError(f"No frames in {frames_dir}")

        ids = uniform_sample_indices(len(frame_paths), self.clip_len)

        clip = []
        for i in ids:
            img = Image.open(frame_paths[i]).convert("RGB")
            clip.append(self.transform(img))

        clip = torch.stack(clip, dim=0)
        return clip, label
```
